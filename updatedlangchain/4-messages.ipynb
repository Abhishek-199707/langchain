Messages

Messages are the basic unit of context for models in LangChain. They are what goes into a model and what comes back out. Each one carries content plus the extra information needed to track the state of a conversation with an LLM.

Every message object contains:

Role: the message type (system, user, assistant, tool).
Content: the text, images, audio, documents, etc.
Metadata: optional fields like response info, message IDs and token usage.
Message types

SystemMessage: tells the model how to behave and gives it context (instructions, persona, rules).
HumanMessage: the user's input to the model.
AIMessage: the model's response, which can include text, tool calls and metadata.
ToolMessage: the result of a tool call, sent back to the model.

```python
import os
from langchain.chat_models import init_chat_model

os.environ["OPENAI_API_KEY"]= os.getenv("OPENAI_API_KEY")
model= init_chat_model("gpt-4.1")
```

```python
model.invoke("Please tell me what is ai?")
```

Message prompts

Alternatively, you can pass messages to the models by providing a list of message objects.

message types
    1. Systems message ; tell model how to behave and provide context fot interaction.
    2. Human message: - Represent user input and interation with the model.
    3. AI message: Response generated from the model including text content , metadata and tool calls.
    4. Tool message - Represents the outputs of tool calls.


```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage
ai_msg = AIMessage("I'd happy to help you with the question!")
messages=[
    SystemMessage("You are a helpful assistant."),
    HumanMessage("Can you help me?"),
    ai_msg,
    HumanMessage("Great! What is 2+2")
]

response=model.invoke(messages)
print(response.content)
```
