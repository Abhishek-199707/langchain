Streaming

Most models can stream their output content while it is being generated. By displaying output progressively streraming significantly improves the user experience , particularly for longer responses. Calling stream() returns an iterator that yields output chunks as they are produced. You can use a loop to process each chunk in real time:

```python
model.invoke("Write me a 200 words paragraph ion artificial intelligence.")
```

```python
for chunk in model.stream("Write me a 200 words paragraph ion artificial intelligence."):
    print(chunk.text, end="|", flush=True)
```

Batch

Batching a collection of independent requests to a modle can significantly improve performance and reduce costs, as the processing can be done in parallel.

```python
from langchain_openai import ChatOpenAI
from langchain.chat_models import init_chat_model
model= init_chat_model("gpt-4.1")
responses = model.batch([
    "Why do parrots have colourful feathers?",
    "How do airplanes fly?",
"does Cetpa infotech is good company?"
],
config={
    "max_concurrency": 5, #Limit to 5 parallel calls.   
}
)

for response in responses:  
    print(response)
```
